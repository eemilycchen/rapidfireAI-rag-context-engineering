```python
import pandas as pd
import glob

pd.set_option('display.max_colwidth', None)
pd.set_option('display.max_columns', None)
```

```python
files = glob.glob('./*.csv')

df_list = [pd.read_csv(f) for f in files]
```

```python
results_df = pd.concat(df_list, ignore_index=True)

results_df = results_df.sort_values(by=['Correctness (pass rate)', 'Faithfulness (pass rate)', 
                                        'F1@5', 'Precision@5', 'Recall@5', 'Retrieval Score'], ascending=False)

results_df.head(5)
```

```python
print(len(results_df))
print('-' * 100)
print(results_df['model_name'].value_counts())
print('-' * 100)
print(results_df['chunk_size'].value_counts())
print('-' * 100)
print(results_df['embedding_cfg'].value_counts())
print('-' * 100)
print(results_df['search_cfg'].value_counts())
```

```python
import ast

def parse_cfg(x):
    if isinstance(x, dict):
        return x
    if isinstance(x, str):
        return ast.literal_eval(x)
    return {}
```

```python
search_cfg = results_df['search_cfg'].apply(parse_cfg)
embedding_cfg = results_df['embedding_cfg'].apply(parse_cfg)
reranker_cfg = results_df['reranker_cfg'].apply(parse_cfg)

results_df['search_type'] = search_cfg.apply(lambda x: x.get('type'))
results_df['search_k'] = search_cfg.apply(lambda x: x.get('k'))
results_df['embedding_class'] = embedding_cfg.apply(lambda x: x.get('class'))
results_df['embedding_model'] = embedding_cfg.apply(lambda x: x.get('model'))
results_df['reranker_class'] = reranker_cfg.apply(lambda x: x.get('class'))
results_df['reranker_model'] = reranker_cfg.apply(lambda x: x.get('model_name'))

results_df.drop(['search_cfg', 'embedding_cfg', 'reranker_cfg'], axis=1, inplace=True)

# different model everything else same
results_df[
    (results_df['chunk_size'] == 512) & 
    (results_df['search_type'] == 'similarity') &
    (results_df['search_k'] == 10) &
    (results_df['embedding_class'] == 'OpenAIEmbeddings') &
    (results_df['reranker_class'] == 'CrossEncoderReranker') &
    (results_df['reranker_model'] == 'cross-encoder/ms-marco-MiniLM-L6-v2')
]
```
