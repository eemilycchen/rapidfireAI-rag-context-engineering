```python
import pandas as pd
import glob

pd.set_option('display.max_colwidth', None)
pd.set_option('display.max_columns', None)
```

```python
files = glob.glob('./*.csv')

df_list = [pd.read_csv(f) for f in files]
```

```python
results_df = pd.concat(df_list, ignore_index=True)

results_df = results_df.sort_values(by=['Correctness (pass rate)', 'Faithfulness (pass rate)', 
                                        'F1@5', 'Precision@5', 'Recall@5', 'Retrieval Score'], ascending=False)

results_df.head(5)
```

```python
print(len(results_df))
print('-' * 100)
print(results_df['model_name'].value_counts())
print('-' * 100)
print(results_df['chunk_size'].value_counts())
print('-' * 100)
print(results_df['embedding_cfg'].value_counts())
print('-' * 100)
print(results_df['search_cfg'].value_counts())
```
